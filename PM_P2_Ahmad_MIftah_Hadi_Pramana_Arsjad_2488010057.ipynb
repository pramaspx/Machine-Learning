Refleksi

1. Apa perbedaan supervised, unsupervised, dan reinforcement learning?
2. Mengapa pembagian data latih dan uji penting?
3. Mengapa alur kerja ML disebut iteratif?


Jawaban
1.
- Supervised learning dimana data diberikan dengan label dan belajar dengan memetakan fitur ke jawaban.
- Unsupervised learning dimana data diberikan tanpa label dan belajar untuk menemulan struktur tersembunyi.
- Reinforcement learning dimana pemeblajaran dilakukan dengan interaksi langsung dengan memberikan reward dan punishment.
2. Pembagian data latih dan data uji penting, karena data latih digunakan untuk pelatihan dengan diberikan contoh dan data latih digunakan untuk pembelajaran agar mendapat hasil dari data.
3. Alur machine learning disebut iteratif karena proses jalannya tidak akan selesasi dalam satu kali jalan, siklusnya akan berjalan terus-menerus untuk meningkatkan akurasi model.
